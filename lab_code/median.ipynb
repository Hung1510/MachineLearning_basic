```python
fileName = input("Enter filename: ")
f = open(fileName,'r')

#input text, convert to number, add the number to a list
numbers =[]
for line in f:
    words = line.split()
    for word in words:
        numbers.append(float(word))

# sort list and print the number at midpoint
numbers.sort()
midpoint = len(numbers) //2

print("median is" , end= " ")
if len(numbers) % 2 == 1:
    print(numbers[midpoint])
else:
    print((numbers[midpoint] + numbers[midpoint -1])/2)
```
