```python
# BARPLOT
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
import pandas as pd
data = pd.read_csv("ipl.csv")
soldprice_by_age =data.groupby("AGE")["SOLD PRICE"].mean().reset_index()
sn.barplot(x='AGE', y='SOLD PRICE', data =soldprice_by_age)
plt.show
```

```python
# BARPLOT

import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
import pandas as pd
data = pd.read_csv("ipl.csv")
soldprice_comparison =data.groupby(
    ['AGE', 'PLAYING ROLE']
)['SOLD PRICE'].mean().reset_index()
sn.barplot(
x='AGE',
y='SOLD PRICE',
hue='PLAYING ROLE',
data=soldprice_comparison
)
plt.show()
```

```python
# HISTPLOT

import matplotlib.pyplot as plt
import pandas as pd

data =pd.read_csv("ipl.csv")

plt.hist(data['SOLD PRICE'], bins=10, edgecolor='black')
plt.xlabel("Sold Price")
plt.ylabel("Number of Players")
plt.title("Distribution of IPL Sold Prices")
plt.show()
```

```python
import pandas as pd
import seaborn as sn
import matplotlib.pyplot as plt

# Load your data
ipl_auction_df = pd.read_csv('ipl.csv')  # adjust path/filename to your file

# Your plot
sn.boxplot(x='PLAYING ROLE', y='SOLD PRICE', data=ipl_auction_df)
plt.figure(figsize=(10, 6))
sn.boxplot(x='PLAYING ROLE', y='SOLD PRICE', data=ipl_auction_df)
plt.xticks(rotation=45)
plt.title('Sold Price Distribution by Playing Role')
plt.tight_layout()
plt.show()
```

```python
# scatter plot
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
import pandas as pd

ipl_auction_df = pd.read_csv("ipl.csv")

ipl_batsman_df = ipl_auction_df[ipl_auction_df['PLAYING ROLE'] == 'Batsman']

plt.scatter(
    x=ipl_batsman_df['SIXERS'],
    y=ipl_batsman_df['SOLD PRICE']
)

plt.xlabel("Sixers")
plt.ylabel("Sold Price")
plt.title("IPL Batsman: Sixers vs Sold Price")
plt.show()
```

```python
#pairplot
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
import pandas as pd

ipl_auction_df = pd.read_csv("ipl.csv")

influential_features = ['SR-B','AVE','SIXERS','SOLD PRICE']

sn.pairplot(ipl_auction_df[influential_features], height=2)
plt.show()
```

```python
# correlation and heatmap

import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
import pandas as pd

ipl_auction_df = pd.read_csv("ipl.csv")

influential_features = ['SR-B','AVE','SIXERS','SOLD PRICE']

#Correlation matrix
corr_matrix = ipl_auction_df[influential_features].corr()

#heatmap
sn.heatmap(corr_matrix, annot=True)

plt.title("Correlation Heatmap of IPL features")
plt.show()
```
