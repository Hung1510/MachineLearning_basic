```python
#question 7
def even_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(even_odd(8))
```

```python
#question 8
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
```

```python
#question 14
def count_even(lst):
    count = 0
    for x in lst:
        if x % 2 == 0:
            count += 1
    return count

print(count_even([1, 2, 5, 6, 8, 9 ,10])) #4
```

```python
#question 28
import matplotlib.pyplot as plt

def average_marks_bar(student_marks):
    names = list(student_marks.keys())
    averages = [sum(marks) / len(marks) for marks in student_marks.values()]
    plt.bar(names, averages, color='red')
    plt.title("Average Marks of Students")
    plt.xlabel("Student")
    plt.ylabel("Average Marks")
    plt.show()

marks_data = {
    'Fangyi': [99, 90, 89],
    'Guiliman': [72, 65, 80],
    'Dorn': [88, 92, 95],
    'Leman': [60, 70, 55]
}
average_marks_bar(marks_data)
```
